# Segmenting `fish.tif` using Python

> This proposed solution is inspired from our guide: [Introduction to Python for Image Analysis](https://epfl-center-for-imaging.github.io/python-intro-images/lab/index.html?path=Content.ipynb).

In the cell below, we read the image into Python, apply a few operations to segment the fish using [Scikit-image](https://scikit-image.org) and [Scipy](https://scipy.org/) functions. We then save the mask as a TIFF image.

```python

from pathlib import Path
from scipy.ndimage import binary_fill_holes
import skimage.io
from skimage.morphology import remove_small_objects
from skimage.measure import label

image_path = Path(".").resolve().parents[1] / "public" / "Images" / "fish.tif"

image = skimage.io.imread(image_path)

binary_img = image < 200
binary_img = binary_fill_holes(binary_img)
binary_img = remove_small_objects(binary_img, min_size=500)

label_img = label(binary_img)

mask_path = Path(".").resolve().parents[1] / "Submissions" / "fish" / "example_solution.tif"

skimage.io.imsave(mask_path, label_img)
```
